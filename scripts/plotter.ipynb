```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

```

```python
path = '../instances/created-instances/'
arquivo = 'mm-5-10-1110588.cdop'
pos = {}
m = 0
n = 0
with open(f'{path}{arquivo}', 'r') as f:
    linhas = f.readlines()
    m = int(linhas[0].split()[1])
    n = int(linhas[1].split()[1])
    
    for i in range(m+n+1):
        x, y = map(float, linhas[i + 2].split()[1:3])
        pos[i] = (x, y)
```

```python
def show_graph(g, pos):
    print(f'g = {g}')
    G = nx.from_numpy_array(g)
    #pos = nx.spring_layout(G)  # layout
    nx.draw(G, pos, with_labels=True, node_color='skyblue', edge_color='gray')

    # Pesos das arestas
    edge_labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

    plt.show()
```

# Truck Graph

```python
g = []
with open(f'{path}{arquivo}', 'r') as f:
    linhas = f.readlines()

    i = 0
    for (i, l) in enumerate(linhas):
        if "Truck Graph:" in l:
            break
    for j in range(m+1):
        g.append(linhas[i + j + 1].split())

g = np.array(g).astype(float)
```

```python
show_graph(g, pos)
```
